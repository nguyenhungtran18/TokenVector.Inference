```python
# 2/3: clone + symlink CUDA
import os, subprocess, glob
d='/kaggle/working/TokenVectorCuda'
subprocess.run('rm -rf '+d, shell=True)
r=subprocess.run('git clone --depth 1 --branch prototype/cuda-backend https://github.com/nguyenhungtran18/TokenVector.git '+d, shell=True, capture_output=True, text=True, timeout=600)
print('clone rc=', r.returncode)
print('HEAD', subprocess.check_output('git -C '+d+' rev-parse --short HEAD', shell=True).decode().strip())
libdir='/kaggle/working/tkvlib'; os.makedirs(libdir, exist_ok=True)
def pick(pats):
    for pat in pats:
        f=[p for p in glob.glob(pat) if os.path.exists(os.path.realpath(p))]
        if f: return sorted(f)[0]
    return None
libs={'cudart.dll':pick(['/usr/local/cuda*/lib64/libcudart.so','/usr/lib/x86_64-linux-gnu/libcudart.so*']),
      'cublas.dll':pick(['/usr/local/cuda*/lib64/libcublas.so','/usr/lib/x86_64-linux-gnu/libcublas.so*']),
      'cuda.dll':pick(['/usr/lib/x86_64-linux-gnu/libcuda.so*','/usr/local/nvidia/lib64/libcuda.so*','/usr/local/cuda/compat/libcuda.so*'])}
for nm,tgt in libs.items():
    print(nm,'->',tgt)
    if tgt:
        p=libdir+'/'+nm
        if os.path.lexists(p): os.remove(p)
        os.symlink(tgt,p)
os.environ['LD_LIBRARY_PATH']=libdir+':'+os.environ.get('LD_LIBRARY_PATH','')
```

```python
# 3/3: chay + so sanh
import subprocess, os, json
exe=d+'/Release/cuda_backend/examples/ilgpt2_trainN.exe'
print('exe bytes=', os.path.getsize(exe) if os.path.exists(exe) else 0)
cfg={'runtimeOptions':{'tfm':'net10.0','framework':{'name':'Microsoft.NETCore.App','version':'10.0.0'}}}
open(os.path.dirname(exe)+'/ilgpt2_trainN.runtimeconfig.json','w').write(json.dumps(cfg))
r=subprocess.run(['/kaggle/working/dotnet/dotnet', exe], env=dict(os.environ), capture_output=True, text=True, timeout=600)
print('dotnet rc=', r.returncode)
print('STDERR:', (r.stderr or '')[:800])
print((r.stdout or '')[:200])
print('...')
keep=[l for l in (r.stdout or '').splitlines() if any(k in l for k in ('LOSS','M CE','M FN','FAIL','MS_','MARK FWD','MARK BWD','MARK UPD'))]
print('KEY LINES:'+"\n"+'\n'.join(keep))
exec(open(d+'/Release/cuda_backend/examples/ref_gpt2_trainN.py').read())
```
